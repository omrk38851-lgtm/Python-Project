```python
%matplotlib inline
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
df=pd.read_csv('/content/Attrition.csv')
df
```

#.what is effect of age on attrition ?

```python
df["Attrition"] = df["Attrition"].map({"Yes":1, "No":0})

age_attrition = df.groupby("Age")["Attrition"].mean()

plt.figure(figsize=(10,6))
plt.plot(age_attrition.index, age_attrition.values, marker='o')
plt.title("Attrition Rate by Age")
plt.xlabel("Age")
plt.ylabel("Attrition Rate")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
#Younger people are more likely to quit.
```

#Is income the main factor towards employee attrition?

```python
df['income_group']=pd.qcut(df["MonthlyIncome"],5)
income_attrition=df.groupby("income_group")["Attrition"].mean()
plt.figure(figsize=(10,6))
plt.plot(income_attrition.index.astype(str),income_attrition.values,marker='o')
plt.title("Attrition Rate by Income")
plt.xlabel("Income Group")
plt.ylabel("Attrition Rate")
plt.xticks(rotation=45)
plt.grid(True,alpha=0.3)
plt.show()
```

```python
#🔹 Low income is associated with high Attrition.
```

#.Does the Department of work impact attrition?

```python
departmet_attrition=df.groupby("Department")["Attrition"].mean()

plt.figure(figsize=(10,6))
plt.bar(departmet_attrition.index,departmet_attrition.values,color=["blue",'red','green'])
plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition")
plt.grid(True,alpha=0.5)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#🔹 The sales department has the highest Leave rate.
```
